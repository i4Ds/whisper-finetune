```python
from datasets import load_dataset
import IPython.display as ipd
import torchaudio
import torch
from torch_audiomentations import *
```

```python
ds = load_dataset('i4ds/sds-stt-spc-train-fold_0', split='train')
```

```python
example = torch.tensor(ds[0]['audio']['array'])
```

```python
example.unsqueeze(0).unsqueeze(0)
```

```python
ipd.Audio(example, rate=16000)  
```

```python
laudio = LowPassFilter(p=1, min_cutoff_freq=3340, max_cutoff_freq=7500, sample_rate=16000)(example.unsqueeze(0).unsqueeze(0))
```

```python
ipd.Audio(laudio.squeeze().squeeze(), rate=16000)
```

```python
haudio = HighPassFilter(p=1, min_cutoff_freq=20, max_cutoff_freq=128, sample_rate=16000)(laudio)
```

```python
ipd.Audio(haudio.squeeze().squeeze(), rate=16000)  
```

```python
naudio = AddColoredNoise(p=1, sample_rate=16000, min_snr_in_db=25, max_snr_in_db=99)(haudio)
```

```python
ipd.Audio(naudio.squeeze().squeeze(), rate=16000)  
```

```python
baudio = BandPassFilter(p=1, sample_rate=16000, min_bandwidth_fraction=1.5)(naudio)
```

```python
ipd.Audio(baudio.squeeze().squeeze(), rate=16000)  
```
